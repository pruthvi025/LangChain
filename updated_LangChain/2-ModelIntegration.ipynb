### we are integrating model like openai, groq,gemini

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
```

### open ai model calling

```python
from langchain.chat_models import init_chat_model
model=init_chat_model("gpt-4.1")
model
```

```python
resopnse=model.invoke("hello, how are you")

resopnse.content
```

### Google gemini model calling


```python
from langchain.chat_models import init_chat_model
import os

os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")

model=init_chat_model("google_genai:gemini-2.5-flash")

resopnse=model.invoke("why do parrots talk??")

resopnse.content
```

### Groq Model Integration



```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")

model=init_chat_model("groq:qwen/qwen3-32b")
response=model.invoke("why do parrots talk??")

response
```

### Streaming Responses


```python
for chunk in model.stream("why do parrots talk??"):
    print(chunk.content, end="", flush=True)
```

### Batch Processing


```python
responses = model.batch(["why do parrots talk??", "what is the capital of France??"])
for res in responses:
    print(res.content)
```
